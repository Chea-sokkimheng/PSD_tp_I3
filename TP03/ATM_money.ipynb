```python
balance = 2490
bill_in_machine = {100: 50, 50: 20, 20: 300, 10: 0, 5: 200, 1: 10}   #100$ has 50 bills, 50$ has 20 bills, 20$ has 300 bills, 10$ has 0 bills, 5$ has 200 bills, and 1$ has 10 bills

print("=== Welcome to the ATM ===")
while True:

    try: 
        amount = int(input("Please input your withdrawal amount: "))
    except ValueError:
        print("Invalid input. Please enter a valid number.")
        continue

    if amount <= 0 or amount > balance:
        print("Invalid amount. Please enter a valid amount.")
        continue
    left = amount
    bill_given = {}
    for bill in sorted(bill_in_machine, reverse=True):   # left // bill is how many of that bill would fit into the remaining amount.
        count = min(left // bill, bill_in_machine[bill]) # min(..., bill_in_machine[bill]) stops it from using more bills than the machine holds.
        if count > 0:
            bill_given[bill] = count
            left -= bill * count

    if left != 0:
            print("Cannot dispense the requested amount.Try another amount.")
            continue
    break  #valid transaction, exit the loop

for bill, count in bill_given.items():
    bill_in_machine[bill] -= count
balance -= amount

# Print receipt
print("Transaction Successful!")
print(f"You have received ${amount}")
for bill, count in bill_given.items():
    print(f"{count} of ${bill} bill")
print(f"Your remaining balance: ${balance}")
print("Remaining in this machine:")
for bill in sorted(bill_in_machine, reverse=True):
    print(f"${bill} bill cash: {bill_in_machine[bill]} ")
```

* balance: is the customer’s account balance.
* bill_in_machine: maps each bill’s value to how many of that bill the machine holds. For example, it has fifty $100 bills and no $10 bills.
